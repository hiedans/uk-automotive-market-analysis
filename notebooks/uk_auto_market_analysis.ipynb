# UK Automotive Market: Electrification & Chinese New Entrants
### A public-data market analysis, built for a data analyst portfolio

**Author:** Hieda Silva

This notebook analyses two structural shifts in the UK new car market:
1. How fast BEV adoption is closing on the legally binding **ZEV Mandate** target
2. How quickly Chinese-origin brands have grown from a standing start in 2022

All figures are compiled from **public sources only** — UK government (DfT/DVLA)
official statistics under the Open Government Licence, and public SMMT-sourced
press reporting. No proprietary or confidential data is used anywhere in this
project. Full source list in `README.md`.


## 1. BEV share vs. the ZEV Mandate

The UK ZEV Mandate legally requires an escalating share of new car sales to be zero-emission each year, reaching 80% by 2030. How does the actual market compare?

**Reading the chart:** actual BEV share tracked the mandate closely through 2022-2023,
but has fallen visibly behind from 2024 onwards — actual 2025 share (23.4%) vs. the
2025 target (28%), and Q1 2026 (21.8%) vs. the 2026 target (33%). This is a known,
publicly reported industry-wide compliance gap, not a single manufacturer's problem.

## 2. Chinese-origin brands: from zero to a top-tier market force

MG (Chinese-owned since 2007) is excluded from the 'new entrants' series since it has been established in the UK market for decades — the interesting story is the wave of brands that launched from 2022 onwards (GWM, BYD, Chery Group's Omoda/Jaecoo, Leapmotor, XPeng).

**Reading the chart:** new-entrant Chinese brands (excluding MG) grew from a negligible
share in 2022 to roughly 12% of new registrations by 2026 — a genuinely fast market
entry by historical standards. Including MG, all Chinese-owned brands combined are
approaching one in six new cars sold in the UK.

## 3. Why this matters for a UK automotive business

- **Compliance risk is real and quantifiable**: any manufacturer benchmarking itself against
  the ZEV Mandate needs a live, not annual, view of where the whole market — not just their
  own brand — sits against the legal curve.
- **New entrants are not a rounding error**: a combined ~12-16% share within four years of
  market entry is fast enough that "monitor competitors" needs to mean monthly tracking, not
  an annual review.
- **A live dashboard beats a point-in-time analysis**: both of these trends move quarter to
  quarter. See the companion Power BI dashboard in `/dashboard` for an operational view.
